```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# loading the dataset to the pandas dataframe
credit_card_data=pd.read_csv('/content/creditcard.csv.zip')
```

```python
# we have two classes class 0 and class 1
# class 0=legit transaction and class 1 is fraud transaction


# distribution of legit transactions and fraudlent transactions

credit_card_data['Class'].value_counts()
```

```python
# there are 284315 legit transactions and 492 fraud transactions
# so our model will not be able to predict the fraud transactions because
# if we find accuracy on the basis of total data our model will always predit legit transaction
```

```python
# now we are going to separate the data for analysis
# we will make two variable one variable will store legit data
# and 2nd one will store fraud data
legit=credit_card_data[credit_card_data.Class==0]
fraud=credit_card_data[credit_card_data.Class==1]
```

```python
print(legit.shape)
print(fraud.shape)
```

```python
# statistical measures of the data
legit.Amount.describe()
```

```python
fraud.Amount.describe()
```

```python
# we can compare the values for both transactions
credit_card_data.groupby('Class').mean()
```

```python
# as there is a very large legit value so we need to take the sample dataset of legit data
legitSample=legit.sample(n=492)
```

```python
# we will concatnate the two dataset
new_dataset=pd.concat([legitSample,fraud],axis=0)
new_dataset.head()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

```python
# splitting the datasets into features and targets
X=new_dataset.drop(columns='Class',axis=1);
Y=new_dataset['Class']
print(X)
print(Y)
```

```python
# split the data into training  data and test data
# for training the model

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

```python
print(X.shape,X_test.shape,X_train.shape)
```

```python
# model training
model=LogisticRegression()
```

```python
# training the logistic regression model with the training data
model.fit(X_train,Y_train)
```

```python
# to find the performance of our model
# Accuracy
# Accuracy on training data
X_train_prediction=model.predict(X_train)
trainingDataAccuracy=accuracy_score(X_train_prediction,Y_train)


```

```python
print('Accuracy on training data:',trainingDataAccuracy)
```

```python
# accuracy  on testdata
X_test_prediction=model.predict(X_test)
testDataAccuracy=accuracy_score(X_test_prediction,Y_test)
print('Accuracy on test data:',testDataAccuracy)
```
